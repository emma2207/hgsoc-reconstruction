# Make CellRanger input files

Based on one manually prepared CellRanger input config .csv file, we adapt it to all pools.

```python
import os
import pandas as pd
import re
```

```python
PATH = f"{os.getcwd()}/../"
file = PATH + "data/cellranger/cellranger_multi_input.csv"
```

```python
df = pd.read_csv(file, header=None)
```

```python
df
```

```python
gex_fastq_files = [
    "Pool1-GEX-CaseyGreene-11302022",
    "Pool2-GEX-CaseyGreene-11302022",
    "Pool3-GEX-CaseyGreene-12122022",
    "Pool4-GEX-CaseyGreene-12122022",
    "Pool5-GEX-CaseyGreene-01042023",
    "Pool6-GEX-CaseyGreene-01042023",
    "Pool7-GEX-CaseyGreene-01052023",
    "Pool8-GEX-CaseyGreene-01052023",
    "Pool9-GEX-CaseyGreene-01052023",
    "Pool10-GEX-CaseyGreene-01052023"
]
csp_fastq_files = [
    "Pool1-CSP-CaseyGreene-11302022",
    "Pool2-CSP-CaseyGreene-11302022",
    "Pool3-CSP-CaseyGreene-12122022",
    "Pool4-CSP-CaseyGreene-12122022",
    "Pool5-CSP-CaseyGreene-01042023",
    "Pool6-CSP-CaseyGreene-01042023",
    "Pool7-CSP-CaseyGreene-01052023",
    "Pool8-CSP-CaseyGreene-01052023",
    "Pool9-CSP-CaseyGreene-01052023",
    "Pool10-CSP-CaseyGreene-01052023"
]
```

```python
output_path = f"{os.getcwd()}/../data/cellranger_input"
os.mkdir(output_path)
```

```python
# Create cellranger input files for every sample pool
for gex, csp in zip(gex_fastq_files, csp_fastq_files):

    # locate rows with fastq files
    fastq_locations = df[df[0].str.startswith("Pool", na=False)].index
    # update fastq file names
    df.loc[fastq_locations[0], 0] = gex
    df.loc[fastq_locations[1], 0] = csp

    # find pool number in file name
    pool = re.search(r"\d{1,2}", gex).group()
    file_name = f"../data/cellranger/cellranger_input/cellranger_multi_input_pool{pool}"
    # write to file
    df.to_csv(file_name, header=None, index=None)
```
