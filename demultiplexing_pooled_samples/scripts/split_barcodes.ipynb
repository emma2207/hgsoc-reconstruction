```python
import os
import pandas as pd
```

```python
def split_barcodes(pool: int, nr_samples=4):
    
    barcode_dir = f"{os.getcwd()}/vireo/pool{pool}/"
    df = pd.read_csv(barcode_dir + "donor_ids.tsv", delimiter="\t")
    df = df[["cell", "donor_id"]]
    
    for sample in range(nr_samples):
    
        df_sample = df[df["donor_id"]==f"donor{sample}"]
        barcode_lst = df_sample["cell"].reset_index(drop=True)
        barcode_filename = f"htseq-count/pool{pool}/sample{sample+1}/barcodes_per_sample.csv"
        barcode_lst.to_csv(barcode_filename, index=False, header=False)
    
    return 
```

```python
# define number of samples per pool
nr_samples_lst = [4 if pool < 9 else (3 if pool < 10 else 2) for pool in range(1, 11)]
```

```python
for pool, nr_samples in zip(range(1, 11), nr_samples_lst):
    print(pool, nr_samples)
    split_barcodes(pool, nr_samples)
```
