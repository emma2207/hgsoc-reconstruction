```python
import pandas as pd
import os
```

```python
metadata_path = os.getcwd() + '/../data/sc_high_grade_glioma_metadata.csv'
```

```python
metadata_df = pd.read_csv(metadata_path)
```

```python
run_name_mapping_df = metadata_df[['Run', 'Sample Name']].drop_duplicates()
```

```python
run_name_mapping_df.to_csv(os.getcwd() + '/../data/sc_high_grade_glioma_run_name_mapping.csv', index=False)
```

```python
dataset = "high_grade_glioma"
datatype = "single_cell"
```

```python
sample_names = run_name_mapping_df["Sample Name"].unique()
```

```python
base_dir = os.getcwd() + '/../data/'
```

```python
# Use dataset and sample names to create a directory structure
def create_output_dirs(base_dir, dataset, sample_names):
    for sample in sample_names:
        dir_path = os.path.join(base_dir, 'fastq', dataset, sample)
        os.makedirs(dir_path, exist_ok=True)
    return 
```

```python
create_output_dirs(base_dir, dataset, sample_names)
```

```python
# Use run-name mapping to move fastq files to the correct location

# Read SRR sample names from directory
srr_files = [f for f in os.listdir(base_dir + 'test_data') if (f.endswith('truncated.fastq.gz') & (f.startswith('SRR')))]
srr_files
```

```python

# Move files to correct location based on mapping
for srr in srr_files:
    run_id = srr.split('_')[0]
    sample_name = run_name_mapping_df.loc[run_name_mapping_df['Run'] == run_id, 'Sample Name'].values
    if len(sample_name) == 1:
        sample_name = sample_name[0]
        dest_dir = os.path.join(base_dir, 'fastq', dataset, sample_name)
        os.rename(os.path.join(base_dir, 'test_data', srr), os.path.join(dest_dir, srr))
    else:
        print(f"Warning: Run ID {run_id} maps to {len(sample_name)} samples.")
```
