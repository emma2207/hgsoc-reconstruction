```python
import pandas as pd

from analysis_shared_functions import (
    calculate_accuracy_metrics_pseudobulk,
    create_pseudobulk_submatrix_vireo,
    parse_heatmap_matrix_vireo,
    parse_sample_matching_results_bamixchecker,
    parse_sample_matching_results_crosscheckfingerprints,
    parse_sample_matching_results_hysys,
    parse_sample_matching_results_ngscheckmate,
)
```

```python
DATA_PATH = "../../data/output_data/"
```

```python
tools = ["BAMixChecker", "CrosscheckFingerprints", "HYSYS", "NGSCheckmate", "Vireo"]
ncells_list = [5000, 10000, 50000, 100000, 500000]
read_depths = [1]
datasets = ["low_grade_glioma"]
```

```python
def loop_accuracy_calculations(DATA_PATH, tools, datasets, ncells_list, read_depths):
    results = []

    for tool in tools:
        for dataset in datasets:
            for ncells in ncells_list:
                for rd in read_depths:
                    print(tool, dataset, ncells, rd)
                    try: 
                        if tool == "BAMixChecker":
                            inferred_matches = parse_sample_matching_results_bamixchecker(DATA_PATH, True, dataset, ncells)
                        elif tool == "CrosscheckFingerprints":
                            inferred_matches = parse_sample_matching_results_crosscheckfingerprints(DATA_PATH, True, dataset, ncells, rd)
                        elif tool == "HYSYS":
                            inferred_matches = parse_sample_matching_results_hysys(DATA_PATH, True, dataset, ncells, rd)
                        elif tool == "NGSCheckmate":
                            inferred_matches = parse_sample_matching_results_ngscheckmate(DATA_PATH, True, dataset, ncells, rd)
                        elif tool == "Vireo":
                            matrix = parse_heatmap_matrix_vireo(DATA_PATH, True, dataset, ncells, rd)
                            inferred_matches = create_pseudobulk_submatrix_vireo(matrix)
                        else: 
                            print(f"Error! Do not recognize tool {tool}")
                            continue
                    except FileNotFoundError:
                        print(f"Could not find data for {tool}, pseudobulk dataset {dataset}, ncells {ncells}, read depth {rd}. ")
                        continue

                    inferred_matches_filtered = inferred_matches.loc[
                        [idx for idx in inferred_matches.index if idx.endswith("_1")],
                        [col for col in inferred_matches.columns if col.endswith("_2")]
                    ]

                    accuracy_metrics = calculate_accuracy_metrics_pseudobulk(inferred_matches_filtered.fillna(0).astype('int64'))

                    result = {
                        "tool": tool, 
                        "pseudobulk": True, 
                        "dataset": dataset, 
                        "ncells": ncells, 
                        "read depth": rd, 
                        "accuracy": accuracy_metrics[0],
                        "balanced_accuracy": accuracy_metrics[1],
                        "precision": accuracy_metrics[2],
                        "recall": accuracy_metrics[3],
                        "f1": accuracy_metrics[4]
                        }
                    results.append(result)

    return pd.DataFrame(results)
```

```python
test = loop_accuracy_calculations(
    DATA_PATH=DATA_PATH,
    tools=tools,
    datasets=datasets,
    ncells_list=ncells_list,
    read_depths=read_depths
    )
```
