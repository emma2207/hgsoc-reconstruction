```python
import pandas as pd
```

```python
DATA_PATH = "/Users/lathouwe/Downloads"
```

```python
read_depth_files = [
    "coverage_central_nervous_system_tumor_bulk.txt",
    "coverage_hgsoc_bulk_chunk_ribo.txt",
    "coverage_hgsoc_bulk_dissociated_polyA.txt",
    "coverage_hgsoc_bulk_dissociated_ribo.txt",
    "coverage_hgsoc_new_bulk_chunk_ribo.txt",
    "coverage_hgsoc_new_bulk_diss_polyA.txt",
    "coverage_hgsoc_single-cell.txt",
    "coverage_high_grade_glioma_bulk.txt",
    "coverage_high_grade_glioma_single-cell.txt",
    "coverage_low_grade_glioma_bulk.txt",
    "coverage_low_grade_glioma_single-cell.txt",
    ]
```

```python
full_df = pd.DataFrame()

for file in read_depth_files:
    df = pd.read_csv(f"{DATA_PATH}/{file}", sep="\t")
    df = df[df["#rname"].str.contains(f"chr\\d", regex=True)]
    mean_depth = df["meandepth"].mean()
    full_df = pd.concat([full_df, pd.DataFrame({
        "file": [file],
        "mean_depth": [mean_depth],
    })], ignore_index=True)

full_df["file"] = full_df["file"].str.replace("coverage_", "").str.replace(".txt", "")
full_df["dataset"] = full_df["file"].str.extract(r"^(.*)_(?:bulk|single-cell)")
full_df["datatype"] = full_df["file"].str.extract(
    r"_(bulk|single-cell|bulk_chunk_ribo|bulk_diss_polyA|bulk_dissociated_ribo|bulk_dissociated_polyA)$"
    )
full_df.sort_values("mean_depth")
```
